### **Outliers Values**

Pada tahap ini, kita mengidentifikasi nilai ekstrem (outliers) pada kolom yang bertipe data numerik. Tujuan dari proses ini adalah untuk memahami seberapa besar distribusi data yang menyimpang dari pola umum, sehingga dapat meningkatkan akurasi analisis dan model yang akan dibangun.

Namun, tidak semua kolom dengan outliers perlu langsung ditangani. Kita tetap harus mempertimbangkan konteks dan makna dari setiap variabel dalam dataset.

Kolom-kolom ini memiliki sejumlah nilai ekstrem yang perlu diperhatikan, terutama jika digunakan dalam pemodelan prediktif.

---

-> Outliers cukup tinggi (> 10%)

* infant deaths (10.72%)

* under-five deaths (13.41%)

* percentage expenditure (13.24%)

* GDP (12.42%)

* Population (10.01%)

* Diphtheria (10.14%)

* HIV/AIDS (18.45%)

* Measles (18.45%)

Kolom-kolom ini menunjukkan distribusi data yang cukup menyebar dan memiliki banyak nilai ekstrem hal ini kemungkinan disebabkan oleh perbedaan kondisi antar negara (misalnya negara maju vs negara berkembang).

```python
results = []

cols = df.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(df))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

# **Eksplorasi Data (EDA)**

## **Comparison/Perbandingan**

Aktivitas: Membandingkan rata-rata Angka Harapan Hidup ('Life expectancy') berdasarkan Status Negara ('Status', yaitu 'Developed' dan 'Developing').

Tujuan: Mengidentifikasi apakah ada perbedaan signifikan dalam angka harapan hidup antara negara maju dan negara berkembang, serta melihat kontribusi masing-masing status terhadap rata-rata harapan hidup.

Visualisasi: Bar Chart (Grafik Batang).

```python
life_expectancy_by_status = df.groupby('Status')['Life expectancy'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(x=life_expectancy_by_status.index, y=life_expectancy_by_status.values, palette='viridis', hue=life_expectancy_by_status.index, legend=False)
plt.title('Average Life Expectancy by Country Status')
plt.xlabel('Status')
plt.ylabel('Average Life Expectancy')
plt.show()
```

**Insight:**

Dari bar chart, terlihat jelas bahwa rata-rata angka harapan hidup di negara maju (Developed) jauh lebih tinggi dibandingkan dengan negara berkembang (Developing), hal ini mengindikasikan adanya kesenjangan yang signifikan dalam kualitas hidup dan sistem kesehatan antara kedua kategori negara tersebut. Negara maju memiliki angka harapan hidup rata-rata sekitar 79.2 tahun, sedangkan negara berkembang hanya sekitar 67.1 tahun, maka dari itu perbedaan ini bisa disebabkan oleh banyak faktor seperti akses terhadap layanan kesehatan, sanitasi, gizi, pendidikan, dan kondisi ekonomi.

## **Composition/Komposisi**

Aktivitas: Melihat komposisi persentase jumlah negara berdasarkan status 'Developing' atau 'Developed'.

Tujuan: Memahami proporsi negara maju dan berkembang dalam dataset, yang dapat mempengaruhi interpretasi analisis secara keseluruhan.

Visualisasi: Pie Chart (Diagram Lingkaran).

```python
status_counts = df['Status'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(status_counts, labels=status_counts.index, autopct='%1.1f%%', startangle=140, colors=['skyblue', 'lightcoral'])
plt.title('Composition of Countries by Status (Developing vs. Developed)')
plt.axis('equal')
plt.show()
```

**Insight:**

Pie chart menunjukkan bahwa sebagian besar negara dalam dataset ini (82.6%) dikategorikan sebagai 'Developing' (negara berkembang), sementara hanya 17.4% yang merupakan 'Developed' (negara maju). Proporsi ini penting untuk dipertimbangkan saat melakukan analisis lebih lanjut, karena dominasi negara berkembang dalam dataset dapat mempengaruhi interpretasi statistik atau model yang dibangun, ini juga menyoroti bahwa masalah-masalah kesehatan dan faktor-faktor yang mempengaruhi harapan hidup di negara berkembang akan memiliki bobot yang lebih besar dalam temuan analisis secara keseluruhan.

## **Distribution/Distribusi**

Aktivitas: Menganalisis distribusi Angka Harapan Hidup ('Life expectancy') secara keseluruhan.

Tujuan: Memahami pola sebaran data angka harapan hidup, mengidentifikasi nilai-nilai umum, rentang, dan apakah ada outlier atau skewness dalam distribusi.

Visualisasi: Histogram dengan KDE (Estimasi Kepadatan Kernel).

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Life expectancy'], bins=20, kde=True)
plt.title('Distribusi Angka Harapan Hidup')
plt.xlabel('Angka Harapan Hidup')
plt.ylabel('Frekuensi')
plt.show()
```

**Insight:**

Histogram distribusi angka harapan hidup menunjukkan pola yang cenderung miring ke kiri (left-skewed), dengan sebagian besar negara memiliki angka harapan hidup yang lebih tinggi, sekitar 70 hingga 80 tahun. Terdapat juga puncak yang lebih kecil pada rentang angka harapan hidup yang lebih rendah (sekitar 50-65 tahun), serta mengindikasikan adanya kelompok negara dengan tantangan kesehatan yang lebih besar, sehingga dua puncak atau distribusi yang lebar menunjukkan variasi yang signifikan dalam harapan hidup antar negara, dari yang sangat rendah hingga sangat tinggi.

## **Relationship/Hubungan**

Aktivitas: Menganalisis hubungan (korelasi) antara Angka Harapan Hidup ('Life expectancy') dan Produk Domestik Bruto ('GDP').

Tujuan: Mengidentifikasi seberapa kuat hubungan antara kondisi ekonomi suatu negara (diwakili oleh GDP) dengan angka harapan hidup penduduknya.

Visualisasi: Heatmap Korelasi.

```python
plt.figure(figsize=(8, 6))
sns.heatmap(data=df[['Life expectancy', 'GDP']].corr(),
            annot=True,
            cmap='viridis',
            fmt='.2f')
plt.title('Korelasi antara Life Expectancy dan GDP')
plt.show()
```

**Insight:**

Heatmap korelasi antara 'Life expectancy' dan 'GDP' menunjukkan hubungan positif yang kuat (0.69). Ini berarti semakin tinggi Produk Domestik Bruto (GDP) suatu negara, semakin tinggi pula angka harapan hidup penduduknya. Korelasi ini masuk akal, karena negara dengan ekonomi yang kuat (GDP tinggi) cenderung memiliki sumber daya yang lebih baik untuk investasi di sektor kesehatan, pendidikan, sanitasi, dan infrastruktur, yang pada gilirannya berkontribusi pada peningkatan angka harapan hidup, ini menggarisbawahi bahwa pentingnya faktor ekonomi dalam menentukan kualitas kesehatan suatu populasi.
